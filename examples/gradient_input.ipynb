# Example for marking gradients

```python
import bikipy
from pathlib import Path
```

```python
%matplotlib qt
```

```python
border_img = Path("C:/Users/Can/Projects/Neuroscience/Imen/data/maze.png")
assert border_img.exists(), f"The image file doesn't exist in {border_img}"
border_img = str(border_img)
```

```python
from PIL import Image

img = Image.open(border_img)
```

```python
img
```

```python
gradient_border = bikipy.preferance.border.GradientBorder.many(
    guiding_image=img, n=3
)
```

```python
for border in gradient_border:
    border.plot()
```
