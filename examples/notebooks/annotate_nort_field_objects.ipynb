# NORT Field object annotation

```python
from glob import glob
from pathlib import Path
import pickle

from bikipy.behaviour.nort.experiment import NortObjectField
```

```python
ROOT = Path("C:/Users/Can/Projects/Neuroscience/bikipy/examples/data")
IMG_ROOT = ROOT / "images" / "nort"

T1 = IMG_ROOT / "before"
T2 = IMG_ROOT / "after"
```

```python
nort_fields = []
```

```python
for i, (habit, novelty) in enumerate(
    zip(glob(str(T1 / "habit*")), glob(str(T1 / "novelty*"))), start=1
):
    nort_fields.append(
        NortObjectField.from_images(i, habit, novelty)
    )
```

```python
with open(ROOT / "t1_labels.pickle", "wb") as outpickle:
    pickle.dump(nort_fields, outpickle)
```
