```python
only_one = False
area_info = None
data_dict = {}
for subdir in os.listdir(str(DATA_DIR)):
    current_border_root_dict = exp_id_vs_areas[subdir.split("_")[1].lower()]
    border_infos = tuple(current_border_root_dict.values())
    if len((ranges := tuple(current_border_root_dict.keys()))) == 1:
        area_info = border_infos[0]
        only_one = True

    data_dict[subdir] = {}
    for file_path in glob(os.path.join(str(DATA_DIR / subdir), "**.h5")):
        exp_id = int(EXP_ID_FINDER.findall(Path(file_path).stem)[0])
        print(exp_id)
        if only_one:
            pass
        elif ranges[0] <= exp_id <= ranges[1]:
            area_info = border_infos[0]
        else:
            area_info = border_infos[1]

        length_unit_per_pixel = 8.0 / np.linalg.norm(
            area_info["center"].base_b - area_info["center"].base_a
        )

        assert area_info

        dlc_data = DeepLabCutReader.from_hdf(
            file_path, (640, 480), midpoint_groups=(("left_ear", "right_ear"),)
        )
        data_dict[subdir][exp_id] = YMaze(
            dlc_data["mid-left_ear-right_ear"],
            area_info["borders"],
            area_info["center"],
            14,
            length_unit_per_pixel,
            exp_id,
        )
    only_one = False
    data_dict[subdir] = dict(
        sorted(data_dict[subdir].items(), key=lambda item: item[0])
    )
```

```python
before = YMaze.export_to_dataframe(data_dict["0_before"].values())
after = YMaze.export_to_dataframe(data_dict["1_after"].values())
```

```python
with pd.ExcelWriter(str(WORKING_DIR / "y_maze" / "phd.ods")) as writer:
    before.to_excel(writer, sheet_name="Before")
    after.to_excel(writer, sheet_name="After")
```
